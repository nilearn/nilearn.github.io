```python
from nilearn.plotting import plot_surf_stat_map, show
from nilearn.datasets import load_fsaverage_data

surf_img = load_fsaverage_data(
    mesh="fsaverage5",
    mesh_type="inflated",
    data_type="thickness",
)

fig = plot_surf_stat_map(
    stat_map=surf_img,
    title="fsaverage thickness",
    vmin=0,
    cmap="inferno",
)

show()
```
